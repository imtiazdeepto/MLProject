## Section 3: Clean & Prepare

```python
#remove the id that not important
df=df.drop('id',axis=1)
#rename output name
df=df.rename(columns={'num':'target'})
df.head()
```

```python
def binarization(x):
    if x != 0:
        return 1
    return 0
df['target']=df['target'].apply(binarization)
df.head()
```

```python
#checking the null value of each column
df.isnull().sum()
#only column which have null value
df.isnull().sum()[df.isnull().sum()>0]
```

```python
missing_pct = (df.isnull().sum() / len(df) * 100).round(1)
df.drop(missing_pct[missing_pct>50])
```
